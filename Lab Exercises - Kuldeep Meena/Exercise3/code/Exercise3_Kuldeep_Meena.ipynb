## Exact prompt

Create a seamless 3-second cinematic image-to-video sequence using all three supplied reference images. Start by showing the astronaut portrait clearly and preserve the person's appearance, orange flight suit, and background. During the first second, use a slow, subtle camera push-in. In the second second, transition smoothly from the astronaut portrait to the espresso cup photograph using a soft cross-dissolve; keep the cup, saucer, spoon, and wooden table visually consistent with the reference. During the final second, transition into the supplied blended composite so that visual elements and colours from both source images are visible together. End with a stable frame based on the blended reference. Use natural lighting, realistic photographic detail, smooth transitions, and restrained camera movement. Do not invent extra people, objects, labels, logos, or text. Duration exactly 3 seconds, 16:9 landscape.

## Camera and transition plan

- **0.0–1.0 seconds:** Gentle push-in on the astronaut reference.
- **1.0–2.0 seconds:** Cross-dissolve to the espresso cup reference.
- **2.0–3.0 seconds:** Blend into the supplied composite and hold the final frame.
- **Camera:** Stable framing; no shake or abrupt zoom.
- **Output:** 3 seconds, 16:9 landscape.
