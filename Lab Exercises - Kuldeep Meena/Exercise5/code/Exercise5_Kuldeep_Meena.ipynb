# Exercise 5 — Interactive AI Prototype with Gradio

**Purpose:** Build a small browser-based conversational interface with a title, usage guidance, and sample questions.

This prototype uses a lightweight rule-based response engine. It is self-contained and does not claim to call an external language model.

```python
%pip install -q gradio

import gradio as gr
```

## 1. Define response categories

The response function recognizes common topics from the practical exercises and returns a short explanation or next step.

```python
TOPIC_GUIDE = {
    "pandas": "Pandas provides DataFrames for tabular data. Typical steps include reading data, cleaning missing values, grouping records, and exporting results.",
    "opencv": "OpenCV is used for image processing. For weighted blending, first resize both images to equal dimensions and then call cv2.addWeighted().",
    "wordcloud": "A Word Cloud emphasizes frequent terms in text. Clean empty summaries and remove common stop words before generating it.",
    "gradio": "Gradio wraps a Python function in a browser interface. ChatInterface is a convenient option for a simple conversational demo.",
    "video": "A good image-to-video prompt specifies the reference images, duration, aspect ratio, camera motion, transition order, and constraints.",
    "excel": "Pandas can export a DataFrame to an Excel workbook with DataFrame.to_excel(), usually with index=False."
}

def training_assistant(message, history):
    question = message.strip().lower()

    if not question:
        return "Please enter a question so I can help."
    if any(greeting in question for greeting in ("hello", "hi", "good morning", "good afternoon")):
        return "Hello! I am the GOT-A Lab Helper. Ask me about Pandas, OpenCV, Word Clouds, video prompts, Excel, or Gradio."

    for keyword, answer in TOPIC_GUIDE.items():
        if keyword in question:
            return answer

    if "thank" in question:
        return "You're welcome. Feel free to ask another practical-lab question."

    return (
        "I can help with the topics covered in this practical: "
        "Pandas data analysis, OpenCV image blending, Word Cloud generation, "
        "AI video prompt design, Excel export, and Gradio interfaces. "
        "Try one of the sample questions below."
    )
```

## 2. Configure the browser interface

The title, description, example questions, and input placeholder make the application easier to test.

```python
chat_app = gr.ChatInterface(
    fn=training_assistant,
    title="GOT-A Practical Lab Support Desk",
    description=(
        "Welcome. This demonstration answers introductory questions about "
        "the Python and AI topics used in the five lab exercises."
    ),
    examples=[
        "How can I calculate a row-wise mean with Pandas?",
        "Why do two images need the same dimensions before OpenCV blending?",
        "What is a Word Cloud used for?"
    ],
    chatbot=gr.Chatbot(height=430),
    textbox=gr.Textbox(
        placeholder="Enter a practical-lab question...",
        lines=1
    )
)
```

## 3. Launch and test

Run this cell and open the local or share URL printed by Gradio. Test at least two sample questions and capture a screenshot for the submission.

```python
chat_app.launch(share=True)
```
