# Exercise 2 — Computer Vision with OpenCV: Weighted Image Composition

**Purpose:** Read two image files, normalize their dimensions, combine them using OpenCV's weighted-addition operation, and save evidence of the processing.

```python
from pathlib import Path
import cv2
import matplotlib.pyplot as plt

INPUT = Path("../input")
OUT = Path("../output")
OUT.mkdir(exist_ok=True)

first_file = INPUT / "Exercise2_Image1_Astronaut.jpg.webp"
second_file = INPUT / "Exercise2_Image2_Coffee.jpg.webp"

```

## 1. Read the image files and prepare a common canvas

OpenCV loads colour images in BGR order. Both images are resized to the same width and height before blending.

```python
portrait = cv2.imread(str(first_file), cv2.IMREAD_COLOR)
coffee_scene = cv2.imread(str(second_file), cv2.IMREAD_COLOR)

if portrait is None or coffee_scene is None:
    raise FileNotFoundError("Check that both source images are present in the input folder.")

canvas_size = (600, 400)  # width, height
portrait = cv2.resize(portrait, canvas_size, interpolation=cv2.INTER_AREA)
coffee_scene = cv2.resize(coffee_scene, canvas_size, interpolation=cv2.INTER_AREA)

print("Prepared image dimensions:", portrait.shape, coffee_scene.shape)
```

## 2. Create the composite

The first image contributes 65% and the second image contributes 35%. The weights add up to 1.0.

```python
composite = cv2.addWeighted(portrait, 0.65, coffee_scene, 0.35, 0)

cv2.imwrite(str(OUT / "Exercise2_Rajveer_Blend.png"), composite)
```

## 3. Compare the inputs and result

```python
panels = [
    ("Astronaut source", portrait),
    ("Coffee source", coffee_scene),
    ("65:35 composite", composite)
]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for axis, (caption, frame) in zip(axes, panels):
    axis.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    axis.set_title(caption)
    axis.axis("off")

fig.tight_layout()
fig.savefig(OUT / "Exercise2_Three_Image_Comparison.png", dpi=160, bbox_inches="tight")
plt.show()

print("Saved composite and comparison figure in:", OUT.resolve())
```

## Observation

The source files are preserved in `input/`. The output image is generated by combining corresponding pixels after resizing both sources to the same dimensions.
