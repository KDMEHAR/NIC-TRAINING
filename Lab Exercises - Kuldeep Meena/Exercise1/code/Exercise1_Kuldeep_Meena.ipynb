# Exercise 1 — Exploratory Data Analysis and Excel Reporting

**Purpose:** Create a numerical dataset, inspect the relationship between two features, calculate a row-level average, and export the finished table.

This version uses a local random generator (so it does not change NumPy's global random state), a different sample size, and a two-panel visual check.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

OUT = Path("../output")
OUT.mkdir(exist_ok=True)

```

## 1. Build the sample dataset

The four required fields contain randomly generated measurements. A fixed generator seed makes the result repeatable.

```python
generator = np.random.default_rng(2026)
sample_count = 150

records = generator.uniform(
    low=5,
    high=95,
    size=(sample_count, 4)
)

analysis_df = pd.DataFrame(records, columns=["f1", "f2", "f3", "f4"])
analysis_df.head()
```

## 2. Visual inspection

The scatter chart compares `f2` and `f3`. The histogram gives a quick view of the `f5` values after the average is calculated.

```python
analysis_df["f5"] = analysis_df[["f1", "f2", "f3", "f4"]].mean(axis="columns")

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].scatter(
    analysis_df["f2"], analysis_df["f3"],
    s=34, alpha=0.65, edgecolors="white", linewidths=0.4
)
axes[0].set(title="Feature relationship: f2 vs f3", xlabel="f2 values", ylabel="f3 values")
axes[0].grid(alpha=0.25)

axes[1].hist(analysis_df["f5"], bins=14, edgecolor="white")
axes[1].set(title="Distribution of row averages", xlabel="f5 (row mean)", ylabel="Number of rows")
axes[1].grid(axis="y", alpha=0.25)

fig.tight_layout()
fig.savefig(OUT / "exercise1_analysis.png", dpi=160, bbox_inches="tight")
plt.show()

```

## 3. Review and export the results

```python
print("Dataset dimensions:", analysis_df.shape)
display(analysis_df.head(8))
display(analysis_df[["f1", "f2", "f3", "f4", "f5"]].describe().round(2))

report_file = OUT / "Exercise1_Rajveer_Data.xlsx"
analysis_df.to_excel(report_file, index=False, sheet_name="Feature_Data")
print("Workbook created:", report_file.resolve())
```

## Findings

- The table contains four generated input features and the calculated `f5` row mean.
- The scatter plot shows the distribution of `f2` and `f3`.
- The Excel workbook contains all five required columns.
